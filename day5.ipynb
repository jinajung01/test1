````python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# 1. Matplotlib의 내부 폰트 캐시를 강제로 삭제하고 갱신
# [확인 필요] 캐시가 갱신되어 새로 설치한 'fonts-nanum' 리스트를 확실하게 동기화합니다.
fm._rebuild()

# 2. 나눔바른고딕을 기본 폰트로 명시적 지정
font_path = '/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf'
font_name = fm.FontProperties(fname=font_path).get_name()
plt.rc('font', family=font_name)
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 해결

# 3. 채널별 매출 합계 계산
channel_sales = df.groupby("채널")["매출"].sum().reset_index()
print("--- 채널별 매출 합계 (표) ---")
display(channel_sales)

# 4. 요일별 매출 합계 계산 (월~일 순서 정렬)
weekday_order = ['월', '화', '수', '목', '금', '토', '일']
weekday_sales = df.groupby("요일")["매출"].sum().reindex(weekday_order).dropna().reset_index()
print("\n--- 요일별 매출 합계 (표) ---")
display(weekday_sales)

# 5. 시각화를 위한 서브플롯 생성 및 그래프 그리기
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 5-1. 채널별 매출 막대그래프
sns.barplot(x="채널", y="매출", data=channel_sales, ax=axes[0], hue="채널", palette="Set2", legend=False)
axes[0].set_title("채널별 매출 합계", fontproperties=fm.FontProperties(fname=font_path, size=14))
axes[0].set_xlabel("채널", fontproperties=fm.FontProperties(fname=font_path, size=12))
axes[0].set_ylabel("매출 (원)", fontproperties=fm.FontProperties(fname=font_path, size=12))

# 5-2. 요일별 매출 막대그래프
sns.barplot(x="요일", y="매출", data=weekday_sales, ax=axes[1], hue="요일", palette="Pastel1", legend=False)
axes[1].set_title("요일별 매출 합계", fontproperties=fm.FontProperties(fname=font_path, size=14))
axes[1].set_xlabel("요일", fontproperties=fm.FontProperties(fname=font_path, size=12))
axes[1].set_ylabel("매출 (원)", fontproperties=fm.FontProperties(fname=font_path, size=12))

# 개별 축(Tick) 레이블에도 폰트 명시적 적용
for ax in axes:
    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontproperties(fm.FontProperties(fname=font_path))

# 레이아웃 정렬 및 표시
plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt   # 데이터 시각화를 위한 matplotlib 라이브러리 임포트
import seaborn as sns             # 그래프 디자인을 위한 seaborn 라이브러리 임포트
import matplotlib.font_manager as fm  # 폰트 설정을 위한 폰트 매니저 임포트

# [확인 필요] 나눔바른고딕 폰트 경로 설정 및 폰트 이름 추출
font_path = '/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf'
font_name = fm.FontProperties(fname=font_path).get_name()

# [확인 필요] Matplotlib의 기본 글꼴을 나눔바른고딕으로 설정하여 한글 깨짐 방지
plt.rc('font', family=font_name)
plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지

# 1. 채널별 매출 합계 계산 및 표 출력
channel_sales = df.groupby("채널")["매출"].sum().reset_index()   # 채널별로 매출을 합산한 후 표 형태로 재정렬
print("--- 채널별 매출 합계 (표) ---")                      # 안내 메시지 출력
display(channel_sales)                                        # 채널별 매출 합계 표 출력

# 2. 요일별 매출 합계 계산 및 표 출력 (월요일부터 일요일 순서로 정렬)
weekday_order = ['월', '화', '수', '목', '금', '토', '일']   # 요일 정렬 기준 순서 정의
weekday_sales = df.groupby("요일")["매출"].sum().reindex(weekday_order).reset_index()  # 요일별 매출 합산 및 순서 정렬
print("\n--- 요일별 매출 합계 (표) ---")                      # 안내 메시지 출력
display(weekday_sales)                                        # 요일별 매출 합계 표 출력

# 3. 시각화를 위한 서브플롯(그래프 영역) 생성 (1행 2열 크기)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))               # 14x6 크기의 전체 그래프 틀 생성

# 3-1. 채널별 매출 합계 막대그래프 그리기
sns.barplot(x="채널", y="매출", data=channel_sales, ax=axes[0], hue="채널", palette="Set2", legend=False) # 1번째 영역에 채널별 매출 시각화
axes[0].set_title("채널별 매출 합계", fontproperties=fm.FontProperties(fname=font_path, size=14))  # 그래프 제목 설정
axes[0].set_xlabel("채널", fontproperties=fm.FontProperties(fname=font_path, size=12))            # X축 레이블 설정
axes[0].set_ylabel("매출 (원)", fontproperties=fm.FontProperties(fname=font_path, size=12))        # Y축 레이블 설정

# 3-2. 요일별 매출 합계 막대그래프 그리기
sns.barplot(x="요일", y="매출", data=weekday_sales, ax=axes[1], hue="요일", palette="Pastel1", legend=False) # 2번째 영역에 요일별 매출 시각화
axes[1].set_title("요일별 매출 합계", fontproperties=fm.FontProperties(fname=font_path, size=14))  # 그래프 제목 설정
axes[1].set_xlabel("요일", fontproperties=fm.FontProperties(fname=font_path, size=12))            # X축 레이블 설정
axes[1].set_ylabel("매출 (원)", fontproperties=fm.FontProperties(fname=font_path, size=12))        # Y축 레이블 설정

# 개별 축(Tick) 눈금 레이블에도 나눔폰트 명시적 적용
for ax in axes:                                               # 생성된 모든 그래프를 순회
    for label in ax.get_xticklabels() + ax.get_yticklabels(): # 모든 축의 글자 레이블을 가져옴
        label.set_fontproperties(fm.FontProperties(fname=font_path)) # 폰트 적용

# 레이아웃 간격 자동 정렬 및 그래프 화면 표시
plt.tight_layout()                                            # 그래프 요소 간의 겹침 방지 조절
plt.show()                                                    # 최종 완성된 그래프 출력
````
